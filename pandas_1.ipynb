### 10. Create a Pandas DataFrame from the dictionary of product orders below.

````python
orders = {'Description': ['LUNCH BAG APPLE DESIGN',
  'SET OF 60 VINTAGE LEAF CAKE CASES ',
  'RIBBON REEL STRIPES DESIGN ',
  'WORLD WAR 2 GLIDERS ASSTD DESIGNS',
  'PLAYING CARDS JUBILEE UNION JACK',
  'POPCORN HOLDER',
  'BOX OF VINTAGE ALPHABET BLOCKS',
  'PARTY BUNTING',
  'JAZZ HEARTS ADDRESS BOOK',
  'SET OF 4 SANTA PLACE SETTINGS'],
 'Quantity': [1, 24, 1, 2880, 2, 7, 1, 4, 10, 48],
 'UnitPrice': [1.65, 0.55, 1.65, 0.18, 1.25, 0.85, 11.95, 4.95, 0.19, 1.25],
 'Revenue': [1.65, 13.2, 1.65, 518.4, 2.5, 5.95, 11.95, 19.8, 1.9, 60.0]}
````

````python
orders_df = pd.DataFrame(orders)

orders_df
````

### 11. Calculate the total quantity ordered and revenue generated from these orders.

````python
print("Total quantity:", orders_df["Quantity"].sum())
print("Total revenue:", orders_df["Revenue"].sum())
````

### 12. Obtain the prices of the most expensive and least expensive items ordered and print the difference.

````python
most_expensive = orders_df["UnitPrice"].max()
least_expensive = orders_df["UnitPrice"].min()

print("Most expensive:", most_expensive)
print("Least expensive:", least_expensive)
print("Difference:", most_expensive - least_expensive)
````

## Let's load another dataset for more exercises

````python
# Run this code:
# Looks for the CSV next to the notebook, then one folder up, then in a data folder.
import os

for path in ["Admission_Predict.csv",
             "../Admission_Predict.csv",
             "data/Admission_Predict.csv",
             "../data/Admission_Predict.csv"]:
    if os.path.exists(path):
        admissions = pd.read_csv(path)
        print("Loaded from:", path)
        break
else:
    print("Admission_Predict.csv not found. Current folder:", os.getcwd())
    print("Files here:", os.listdir("."))

# Strip trailing spaces from the column names
# (in this file 'LOR ' and 'Chance of Admit ' have one)
admissions.columns = admissions.columns.str.strip()
admissions.columns.tolist()
````

Let's evaluate the dataset by looking at the `head` function.

````python
admissions.head()
````

### 1 - Before beginning to work with this dataset and evaluating graduate admissions data, we will verify that there is no missing data in the dataset. Do this in the cell below.

````python
# All zeros means nothing is missing
admissions.isnull().sum()
````

###  2 -  Interestingly, there is a column that uniquely identifies the applicants. This column is the serial number column. Instead of having our own index, we should make this column our index. Do this in the cell below. Keep the column in the dataframe in addition to making it an index.

````python
# drop=False keeps "Serial No." as a column as well as making it the index
admissions.set_index("Serial No.", drop=False, inplace=True)
admissions.head()
````

Turns out that `GRE Score` and `CGPA` also uniquely identify the data. Show this in the cell below.

````python
# 0 duplicates means the combination is unique for every row
print(admissions[["GRE Score", "CGPA"]].duplicated().sum())
print(len(admissions[["GRE Score", "CGPA"]].drop_duplicates()), "unique pairs of", len(admissions), "rows")
````

### 3 - In this part of the lab, we would like to test complex conditions on the entire data set at once. Let's start by finding the number of rows where the CGPA is greater than 9 and the student has performed an investigation.

````python
len(admissions[(admissions["CGPA"] > 9) & (admissions["Research"] == 1)])
````

### 4 - Now return all the rows where the CGPA is greater than 9 and the SOP score is less than 3.5. Find the mean chance of admit for these applicants.

````python
rows = admissions[(admissions["CGPA"] > 9) & (admissions["SOP"] < 3.5)]
display(rows)

print("Mean chance of admit:", rows["Chance of Admit"].mean())
````
